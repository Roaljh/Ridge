```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("real_estate_data_updated.csv")
```

```python
numeric_df = df.select_dtypes(include="number")

numeric_features = [col for col in numeric_df.columns if col != "X1 transaction year"]

for col in numeric_features:
    plt.hist(
        df[col].dropna(),
        bins=25,
        edgecolor="black"
    )

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```
