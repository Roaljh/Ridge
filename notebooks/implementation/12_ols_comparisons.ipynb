# Comparative Analysis of Ordinary Least Squares (OLS) Methods

## 1. Overview

This experiment presents a comparative study of three numerical approaches for solving the Ordinary Least Squares (OLS) regression problem:

1. Normal Equations
2. QR Decomposition
3. Singular Value Decomposition (SVD)

The objective is to examine the similarities and differences between these methods in terms of regression performance, coefficient estimation, prediction accuracy, residual behaviour, and numerical stability.

The implementations are presented separately in the following notebooks:

- [09 – OLS using Normal Equations](../implementation/09_ols_normal_equations.ipynb)
- [10 – OLS using QR Decomposition](../implementation/10_ols_qr_decomposition.ipynb)
- [11 – OLS using Singular Value Decomposition](../implementation//11_ols_svd.ipynb)

The numerical outputs generated by these experiments are used for the comparative analysis presented here. No additional regression implementation is performed in this experiment.

---

## 2. Experimental Setup

All three methods use the UCI Real Estate Valuation dataset, consisting of 414 observations.

The revised model specification is used to ensure consistency across all three implementations.

### 2.1 Model Specification

| Component | Description |
|---|---|
| Dataset | UCI Real Estate Valuation |
| Number of observations | 414 |
| Target variable | Y house price of unit area |
| Numerical predictors | House age, MRT distance, convenience stores, latitude, longitude |
| Categorical predictor | Transaction month |
| Encoding | One-hot encoding with `drop_first=True` |
| Intercept | Included |
| Design matrix dimensions | 414 × 17 |

The design matrix contains five numerical predictors, eleven transaction-month indicator variables, and one intercept column.

### 2.2 Comparison Conditions

The same design matrix, target variable, observations, and model specification are maintained across all three methods.

This ensures that differences in the numerical solutions can be examined without introducing variations in the underlying regression model.

## 3. Regression Performance Comparison

The predictive performance of the three OLS approaches is evaluated using four standard regression metrics:

- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- Mean Absolute Error (MAE)
- Coefficient of Determination (R²)

| MAE | 6.013422 | 6.013422 | 6.013422 |
| R² | 0.594059 | 0.594059 | 0.594059 |

### 3.2 Interpretation

All three numerical approaches produce practically identical regression performance.

The MSE, RMSE, and MAE values indicate that the prediction errors are consistent across the three implementations. Similarly, the coefficient of determination is approximately 0.5941 for all methods.

Therefore, for the revised model specification, the choice of numerical solution technique does not produce a meaningful difference in the training-set predictive performance.

However, equivalent predictive performance does not necessarily imply equivalent numerical stability. This distinction is examined in the following sections.

## 7. Numerical Stability Comparison

Numerical stability is an important consideration when solving regression problems, particularly when the design matrix is ill-conditioned.

The numerical properties of the common design matrix are summarized below.

### 7.1 Matrix Properties

| Diagnostic | Value |
|---|---:|
| Number of observations | 414 |
| Number of design matrix columns | 17 |
| Numerical rank | 17 |
| Rank deficiency | 0 |
| Largest singular value | 33867.718371 |
| Smallest singular value | 0.001421 |
| Condition number of X | 2.384011 × 10⁷ |
| Condition number of XᵀX | 5.683462 × 10¹⁴ |

The design matrix has full column rank, indicating that the revised model does not contain an exact linear dependency among its columns.

However, its relatively large condition number indicates that the matrix is still ill-conditioned.

### 7.2 Normal Equations

The Normal Equations approach solves the system:

\[
(X^TX)\hat{\beta}=X^Ty
\]

The condition number of the Gram matrix is approximately the square of the condition number of the design matrix:

\[
\kappa(X^TX)\approx\kappa(X)^2
\]

For the present experiment, the condition number increases from approximately \(2.38\times10^7\) for \(X\) to \(5.68\times10^{14}\) for \(X^TX\).

This indicates that forming the Gram matrix can amplify numerical sensitivity.

Although the revised Normal Equations implementation produces predictions consistent with the other approaches, the method is more vulnerable to numerical errors when the design matrix is ill-conditioned.

### 7.3 QR Decomposition

QR decomposition factorizes the design matrix into:

\[
X=QR
\]

where \(Q\) has orthonormal columns and \(R\) is an upper triangular matrix.

The least-squares solution is obtained by solving:

\[
R\hat{\beta}=Q^Ty
\]

QR decomposition avoids explicitly forming \(X^TX\), thereby avoiding the additional conditioning associated with the Gram matrix.

The numerical verification of the QR decomposition gives:

| Diagnostic | Value |
|---|---:|
| Reconstruction error | 7.325453 × 10⁻¹² |
| Orthogonality error | 2.346880 × 10⁻¹⁵ |
| Numerical rank | 17 |
| Rank deficiency | 0 |

These results indicate that the QR factorization is computed with high numerical accuracy.

### 7.4 Singular Value Decomposition

SVD factorizes the design matrix as:

\[
X=U\Sigma V^T
\]

The least-squares solution for a full-column-rank matrix is:

\[
\hat{\beta}=V\Sigma^{-1}U^Ty
\]

The singular values provide information about the numerical properties of the design matrix and help identify directions associated with small scaling factors.

For the present experiment, the numerical tolerance used in the SVD implementation is:

\[
\tau=3.113340\times10^{-9}
\]

All 17 singular values are greater than this tolerance. Therefore, the numerical rank remains 17.

The tolerance sensitivity experiment also retains all 17 singular values for the tested multipliers of 1, 10, 100 and 1000.

This demonstrates that the revised design matrix is full rank under the tested numerical tolerances.

## 9. Initial and Revised Model: An Additional Observation

An important observation was made during the initial Normal Equations experiment.

The initial model included both transaction year and transaction month as categorical predictors.

However, the observed year-month combinations created an exact linear dependency in the one-hot encoded design matrix.

The initial model therefore contained 18 columns but had a numerical rank of only 17.

### 9.1 Initial vs Revised Model

| Diagnostic | Initial Model | Revised Model |
|---|---:|---:|
| Number of columns | 18 | 17 |
| Matrix rank | 17 | 17 |
| Rank deficiency | 1 | 0 |
| Condition number of X | 9.494351 × 10¹⁸ | 2.384011 × 10⁷ |
| Condition number of XᵀX | 2.855829 × 10²⁰ | 5.683462 × 10¹⁴ |
| MSE | 74.973033 | 74.973033 |
| RMSE | 8.658697 | 8.658697 |
| MAE | 6.013422 | 6.013422 |
| R² | 0.594059 | 0.594059 |

The initial model exhibited rank deficiency because the transaction year and transaction month indicators were not independently estimable under the observed data distribution.

The revised specification removes the transaction year predictor while retaining transaction month as a categorical variable.

This removes the exact linear dependency and produces a full-column-rank design matrix.

Interestingly, the training regression metrics remain unchanged to the displayed precision.

This demonstrates that a model can produce practically identical predictions while having substantially different coefficient identifiability and numerical conditioning.

## 10. Comparative Discussion

The three numerical approaches provide practically identical regression performance for the revised model specification.

However, their numerical characteristics differ.

| Method | Main Advantage | Main Limitation |
|---|---|---|
| Normal Equations | Direct and straightforward formulation | Forming \(X^TX\) can amplify numerical conditioning problems |
| QR Decomposition | Avoids explicitly forming the Gram matrix and provides a stable least-squares solution | Requires matrix factorization and triangular-system solution |
| SVD | Provides singular-value information, numerical rank assessment and pseudoinverse capability | Can involve greater computational cost than simpler approaches |

The experiment highlights that predictive performance alone is insufficient for evaluating numerical solution techniques.

Although all three methods produce practically identical fitted values, QR decomposition and SVD avoid the explicit formation of the Gram matrix and provide more reliable approaches for solving ill-conditioned least-squares problems.

SVD additionally provides a useful framework for examining numerical rank and handling rank-deficient systems through the Moore–Penrose pseudoinverse.

---

## 11. Conclusion

The comparative experiment demonstrates that Normal Equations, QR decomposition and SVD produce practically equivalent OLS regression results for the revised real estate valuation model.

All three methods achieve an MSE of approximately 74.973033, an RMSE of 8.658697, an MAE of 6.013422 and an R² of 0.594059.

Despite this agreement in predictive performance, their numerical characteristics differ.

The Normal Equations approach forms the Gram matrix, whose condition number is significantly larger than that of the original design matrix. QR decomposition avoids this explicit formation, while SVD provides additional insight into the singular-value structure and numerical rank of the system.

The initial and revised model comparison also demonstrates the effect of exact linear dependency among categorical predictors. Removing the redundant transaction-year representation eliminates rank deficiency without changing the reported training metrics.

Overall, QR decomposition and SVD offer more numerically reliable approaches for solving the least-squares problem in the presence of ill-conditioning.

These observations motivate the next experiment: **Ridge Regression**, which introduces regularization to address coefficient instability and multicollinearity.
