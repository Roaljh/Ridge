```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../data/processed/real_estate_data_updated.csv")
```

```python
categorical_features = [
    "X1 transaction year",
    "X1 transaction month"
]

numeric_df = df.select_dtypes(include="number").drop(
    columns=categorical_features,
    errors="ignore"
)
```

```python
outlier_results = []

for col in numeric_df.columns:
    q1 = numeric_df[col].quantile(0.25)
    q3 = numeric_df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = numeric_df[
        (numeric_df[col] < lower) | (numeric_df[col] > upper)
    ]

    print(f"features: {col}")
    print("lower boundary:", lower)
    print("upper boundary:", upper)
    print("outlier count:", len(outliers))

    outlier_results.append({
        "Feature": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Boundary": lower,
        "Upper Boundary": upper,
        "Outlier Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

from pathlib import Path

tables_dir = Path("../../results/data_processing/tables")
outlier_summary.to_csv(tables_dir/"outlier_summary.csv", index=False)
```
