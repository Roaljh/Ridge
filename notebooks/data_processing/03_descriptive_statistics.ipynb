```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../../data/processed/real_estate_data_updated.csv")
```

```python
numeric_df = df.select_dtypes(include=[np.number])
categorical_df = df.select_dtypes(exclude=[np.number])
```

```python
# Descriptive Analysis

print("\n[1] numerical summary")
print(numeric_df.describe().T)

print("\n[2] median")
print(numeric_df.median())

print("\n[3] variance")
print(numeric_df.var())

print("\n[4] standard deviation")
print(numeric_df.std())

print("\n[5] skewness")
print(numeric_df.skew())

print("\n[6] kurtosis")
print(numeric_df.kurtosis())

print("\n[7] categorical summary")
print(categorical_df.describe().T)

print("\n[8] month frequency")
print(categorical_df['X1 transaction month'].value_counts())

print("\n[9] year frequency")
print(numeric_df['X1 transaction year'].value_counts())
```
