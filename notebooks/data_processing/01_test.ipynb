```python
from ucimlrepo import fetch_ucirepo
import numpy as np
import pandas as pd
import calendar
```

```python
dataset = fetch_ucirepo(id=477)

x = dataset.data.features
y = dataset.data.targets

df = pd.concat([x, y], axis=1)

df.to_csv('../../data/raw/real_estate_data.csv', index=False)

print("Data Saved")
print(df.head())
print("Dataset Shape:", df.shape)
```

```python
# 1) Load Dataset

df = pd.read_csv('../../data/raw/real_estate_data.csv')
```

```python
# 2) basic information about the dataset

print("\n[1] Dataset Dimensions:")
print("rows:", df.shape[0])
print("columns:", df.shape[1])

print("\n[2] Column Names:")
print(df.columns.tolist())

print("\n[3] Data Types:")
print(df.dtypes)

print("\n[4] Dataset Info:")
print(df.info())
```

```python
# 3) missing value check

print("\n[5] Missing Values:")
print("total missing values:", df.isnull().sum().sum())

# check empty strings and whitespace strings
empty_strings = {}
for col in df.select_dtypes(include="object").columns:
    empty_strings[col] = (
        df[col].astype("string").str.strip().eq("").sum()
    )

print("\n[6] Empty Strings:")
print(pd.Series(empty_strings))
```

```python
# 4) duplicate check

print("\n[7] Duplicate check:")
print("duplicate rows:", df.duplicated().sum())
```

```python
# 5) type consistency check

print("\n[8] Type Consistency Check:")

for col in df.columns:
    print(f"Column: {col}")
    print(
        df[col].map(
            lambda x: type(x).__name__
        ).value_counts(dropna=False)
    )
```

```python
# 6) numeric conversion check

print("\n[9] Numeric Conversion Check:")

for col in df.columns:
    converted = pd.to_numeric(df[col], errors="coerce")
    invalid = df[col].notna() & converted.isna()
    if invalid.sum() > 0:
        print(f"\n{col}: {invalid.sum()} invalid values")
        print("Examples:", df.loc[invalid, col].head().tolist())
```

```python
# 7) infinite value check

print("\n[10] Infinite Value Check:")

numeric_df = df.select_dtypes(include=np.number)
print(np.isinf(numeric_df).sum())
print(
    "total infinite values:", 
    np.isinf(numeric_df).sum().sum()
)
```

```python
# 8) statistical summary

print("\n[11] Statistical Summary:")
print(df.describe(include="all"))
```

```python
# 9) unique value check

print("\n[12] Unique Value Check:")
for col in df.columns:
    print(f"\n{col}: {df[col].nunique()} unique values")
```

```python
# 10) date column check

print("\n[13] Date Column Check:")
date_cols = [
    col for col in df.columns 
    if "transcation data" in col.lower()
]

if date_cols:
    date_col = date_cols[0]

    dates = pd.to_numeric(df[date_col], errors="coerce")
    
    print("Date Column:", date_col)
    print("min date:", dates.min())
    print("max date:", dates.max())

    print("\nUnique Dates:")
    print(sorted(dates.dropna().unique()))

    fraction = dates - np.floor(dates)
    print("\nFractional year of Dates:")
    print(sorted(fraction.dropna().unique()))
```

```python
# 11) correlation matrix

print("\n[14] Correlation Matrix:")
print(numeric_df.corr())
```
