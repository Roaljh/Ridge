```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../data/raw/real_estate_data.csv")

date_col = "X1 transaction date"

decimal_date = pd.to_numeric(df[date_col], errors="coerce")
year = np.floor(decimal_date).astype(int)
month_number = (
    np.rint((decimal_date - year) * 12 + 1).astype(int)
)

month_name = pd.to_datetime(
    month_number.astype(str), 
    format="%m"
).dt.month_name()

df.drop(
    columns=["X1 transaction year", "X1 transaction month"],
    errors="ignore",
    inplace=True
)

position = df.columns.get_loc(date_col)

df.insert(position, "X1 transaction year", year)
df.insert(position + 1, "X1 transaction month", month_name)

df.drop(columns=[date_col], inplace=True)

df.to_csv("../../data/processed/real_estate_data_updated.csv", index=False)

print(df.head())
```
